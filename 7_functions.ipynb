```python
# Function with Return Value
```

```python
def add(a, b):
    return a + b
```

```python
result = add(10, 20)
```

```python
print(result)
```

```python
def add_mul(a, b):
    add = a + b
    mul = a * b
    return add + mul
```

```python
result = add_mul(10, 10)
```

```python
print(result)
```

```python
# Default Parameter
```

```python
def greet(name="User"):
    print("Hello", name)
```

```python
greet()
```

```python
greet("Rahul")
```

```python
def programming_lang (lang="Python"):
    print(f"I like {lang} programming.")
```

```python
programming_lang()
```

```python
programming_lang("Java")
```

```python
# Multiple Parameters
```

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)
```

```python
student("Rahul", 25)
```

```python
# Types of Arguments
```

```python
# Positional Arguments
```

```python
def show(name, age):
    print(name, age)
```

```python
show("Rahul", 25)
```

```python
# Keyword Arguments
show(age=25, name="Rahul")
```

```python
# Default Arguments
def show(name="Guest"):
    print(name)
```

```python
show()
```

```python
# Variable-length Arguments
```

```python
# *args (multiple values)
```

```python
def add(*nums):
    print(sum(nums))
```

```python
add(10, 20, 30)
```

```python
def multiplication (*arg):
    multi = 1
    for number in arg:
        multi *= number
    return multi
```

```python
multiplication(10,20,30)
```

```python
# **kwargs (key-value)
```

```python
def info(**data):
    print(data)
```

```python
info(name="Rahul", age=25)
```

```python
def add(a, b):
    return a + b
```

```python
result = add(5, 3)
```

```python
print(result)
```

```python
# Lambda Function (Anonymous Function)
```

```python
add = lambda a, b: a + b

print(add(10, 20))
```
