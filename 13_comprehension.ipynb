```python
# Comprehension
# It is a concise way to create collections like list, set, or dictionary using a single line of code with a loop 
# and optional condition.
# Uses loop + condition + expression
# Written in single line
```

```python
# List Comprehension
```

```python
# List comprehension is a short way to create a list using a loop in one line.
```

```python
# [expression for item in iterable if condition]
```

```python
# Basic
```

```python
numbers = [1, 2, 3, 4, 5]
```

```python
squares = [x*x for x in numbers]
```

```python
print(squares)
```

```python
# With Condition
```

```python
numbers = [1, 2, 3, 4, 5, 6]
```

```python
even_numbers = [x for x in numbers if x % 2 == 0]
```

```python
print(even_numbers)
```

```python
# Set Comprehension
```

```python
# Set comprehension is used to create a set using a loop in one line.
```

```python
# {expression for item in iterable if condition}
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5]
```

```python
unique_squares = {x*x for x in numbers}
```

```python
print(unique_squares)
```

```python
# Dictionary Comprehension
```

```python
# Dictionary comprehension is used to create a dictionary (key-value pairs) in one line.
```

```python
# {key_expression: value_expression for item in iterable if condition}
```

```python
numbers = [1, 2, 3, 4]
```

```python
square_dict = {x: x*x for x in numbers}
```

```python
print(square_dict)
```

```python
numbers = [1, 2, 3, 4, 5]
```

```python
even_dict = {x: x*x for x in numbers if x % 2 == 0}
```

```python
print(even_dict)
```

```python
# Generator Comprehension
```

```python
# Generator comprehension creates a generator, not a list.
# It does NOT store all values in memory
# It generates values one by one (lazy evaluation)
```

```python
# (expression for item in iterable if condition)
```

```python
numbers = [1, 2, 3, 4, 5]
```

```python
gen = (x*x for x in numbers)
```

```python
print(gen)
```

```python
for value in gen:
    print(value)
```
