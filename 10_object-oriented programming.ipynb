```python
# Class Methods
```

```python
class EmployeeCM:

    company = "ABC"

    @classmethod
    def change_company(cls, name):
        cls.company = name
```

```python
# Static Methods
```

```python
# Static methods are utility methods that do not use self or cls.
```

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
Calculator.add(10, 20)
```

```python
# Encapsulation
```

```python
# Encapsulation protects data by restricting direct access.
```

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance
```

```python
BankAccount(1000).get_balance()
```

```python
# Property Decorator
```

```python
# Property decorators provide controlled access to attributes.
```

```python
class EmployeeProperty:

    def __init__(self, salary):
        self.__salary = salary

    @property
    def salary(self):
        return self.__salary
```

```python
EmployeeProperty(50000).salary
```

```python
# Single Inheritance
```

```python
# Single inheritance allows one child class to inherit from one parent.
```

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    pass
```

```python
Dog().eat()
```

```python
# Super
```

```python
# super() is used to access parent class methods and constructors.
```

```python
class Parent:
    def __init__(self):
        print("Parent")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child")
```

```python
Child()
```

```python
# Method overriding 
```

```python
# Method overriding replaces a parent method in the child class.
```

```python
class Vehicle:
    def start(self):
        print("Vehicle Start")

class Car(Vehicle):
    def start(self):
        print("Car Start")
```

```python
Car().start()
```

```python
# Polymorphism
```

```python
# Polymorphism allows the same method name to behave differently.
```

```python
class Cat:
    def sound(self):
        print("Meow")

class Cow:
    def sound(self):
        print("Moo")
```

```python
for animal in [Cat(), Cow()]:
    animal.sound()
```

```python
# Abstract classes
```

```python
# Abstract classes define methods that child classes must implement.
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):

    def area(self):
        return 78.5
```

```python
Circle().area()
```

```python
# Multiple inheritance
```

```python
# Multiple inheritance allows a class to inherit from multiple parents.
```

```python
class Father:
    def drive(self):
        print("Driving")

class Mother:
    def cook(self):
        print("Cooking")

class ChildMI(Father, Mother):
    pass
```

```python
c = ChildMI()
```

```python
c.drive()
```

```python
c.cook()
```

```python
# Operator overloading
```

```python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
```

```python
Number(10) + Number(20)
```
