```python
# OOP in Python (Object-Oriented Programming)
```

```python
# OOP is a programming style where we organize code using classes and objects.
```

```python
# Class = blueprint (design)
# Object = real thing created from blueprint
```

```python
# Class
# A Class is a blueprint or a template. 
# It defines the "look" (data) and "actions" (functions) that something should have. 
# It doesn't actually exist as a "thing" yet—it's just the plan.
# Think of: A Recipe for a cake.
```

```python
# Object
# An Object is the actual item created from that blueprint. 
# It is a real "instance" that you can actually use in your code.
# Think of: The Physical Cake you baked using that recipe.
```

```python
# The Key Difference:
# A Class is the idea; an Object is the reality.
```

```python
# Class and Object
```

```python
# Define the 'Car' class (the blueprint)
class Car:
    # The __init__ method (constructor) initializes the object's attributes
    def __init__(self, brand, color):
        self.brand = brand  # Assign the brand name to the object
        self.color = color  # Assign the color to the object

    # Method to display the car's information
    def show(self):
        print("Brand:", self.brand)
        print("Color:", self.color)

# --- Object Creation ---

# Create an instance (object) of the Car class
car1 = Car("Toyota", "Red")

# Call the 'show' method to print the car1 details
car1.show()
```

```python
# self is a reference to the specific object you are currently working with.
# It allows the class to distinguish between its own "blueprint" and the actual "instance" (the real object) that is running the code.
```

```python
# Encapsulation (Data Hiding)
# Wrapping data + methods together in one unit (class)
```

```python
class BankAccount:
    def __init__(self, balance):
        # The double underscore (__) makes 'balance' a private variable
        # It cannot be accessed or changed directly from outside the class
        self.__balance = balance  

    # Method to safely modify the private balance
    def deposit(self, amount):
        self.__balance += amount

    # Method to safely view the private balance (a "Getter")
    def get_balance(self):
        return self.__balance

# Create a bank account object with an initial balance of 1000
acc = BankAccount(1000)

# Use the deposit method to add money
acc.deposit(500)

# Access the balance through the public method
print(acc.get_balance()) 

# Note: Trying to print(acc.__balance) would now result in an Attribute Error!
```

```python
# Inheritance (Code Reusability)
# One class inherits another class
```

```python
# Parent class (Base class)
class Animal:
    def sound(self):
        print("Animal makes sound")

# Child class (Derived class) inheriting from Animal
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Create an object of the Child class (Dog)
d = Dog()

# Calling method from the Parent class (Animal)
# This works because Dog inherited 'sound' from Animal
d.sound()

# Calling method from the Child class (Dog)
d.bark()
```

```python
# Polymorphism (Many Forms)
# Same method name, different behavior
```

```python
# Define a Car class
class Car:
    def move(self):
        print("Car drives on the road")

# Define a Plane class
class Plane:
    def move(self):
        print("Plane flies in the sky")

# Define a Boat class
class Boat:
    def move(self):
        print("Boat sails on the water")

# This function accepts any object and calls its 'move' method
def start_journey(vehicle):
    vehicle.move()

# Creating different objects and passing them to the same function
start_journey(Car())
start_journey(Plane())
start_journey(Boat())
```

```python
# Abstraction (Hiding Implementation)
```

```python
# Show only important details, hide complexity
```

```python
from abc import ABC, abstractmethod

# Define an Abstract Base Class (cannot be used to create objects)
class Shape(ABC):

    # This decorator ensures every child class MUST implement this method
    @abstractmethod
    def area(self):
        pass

# Inherit from Shape; we MUST define 'area' here or Python will throw an error
class Circle(Shape):
    def area(self):
        print("Area of circle")

# Create an instance of the Circle class
c = Circle()

# Call the implemented method
c.area()
```
